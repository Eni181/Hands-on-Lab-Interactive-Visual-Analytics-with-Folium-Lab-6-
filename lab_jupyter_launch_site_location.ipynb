```python
# Task 2: Mark the success/failed launches for each site on the map
import folium
import pandas as pd
from folium.plugins import MarkerCluster
from folium.features import DivIcon

# --- INITIALIZE MAP ---
# Centering on the approximate middle of the US launch sites
nasa_coordinate = [28.562302, -80.577356]
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# --- TASK 1: MARK LAUNCH SITES ---
# Dictionary of unique launch sites and their coordinates
launch_sites_dict = {
    'CCAFS LC-40': [28.562302, -80.577356],
    'CCAFS SLC-40': [28.563197, -80.576820],
    'KSC LC-39A': [28.573255, -80.646895],
    'VAFB SLC-4E': [34.632834, -120.610745]
}

for site, coord in launch_sites_dict.items():
    # Add a Circle for visual area
    circle = folium.Circle(coord, radius=50, color='#d35400', fill=True).add_child(folium.Popup(site))
    
    # Add a text label using DivIcon
    marker = folium.map.Marker(
        coord,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12; color:#d35400;"><b>{site}</b></div>',
        )
    )
    site_map.add_child(circle)
    site_map.add_child(marker)

# --- TASK 2: MARK SUCCESS/FAILURE WITH CLUSTERS ---
# 1. Create a MarkerCluster object
marker_cluster = MarkerCluster()
site_map.add_child(marker_cluster)

# 2. Function to assign marker colors
def assign_marker_color(launch_outcome):
    return 'green' if launch_outcome == 1 else 'red'

# 3. Apply color logic to the dataframe
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# 4. Add each launch record to the marker cluster
for index, record in spacex_df.iterrows():
    marker = folium.Marker(
        location=[record['Lat'], record['Long']],
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=f"Site: {record['Launch Site']} | Outcome: {'Success' if record['class']==1 else 'Failure'}"
    )
    marker_cluster.add_child(marker)

# Display the final map
site_map
```

```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster, MousePosition
from folium.features import DivIcon
from math import sin, cos, sqrt, atan2, radians

# --- 1. HELPER FUNCTION: CALCULATE DISTANCE ---
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0 # Earth radius in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon, dlat = lon2 - lon1, lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

# --- 2. INITIALIZE MAP & PLUGINS ---
nasa_coordinate = [28.562302, -80.577356]
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# Add Mouse Position to help find coordinates for Task 3
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)
site_map.add_child(mouse_position)

# --- 3. TASK 1 & 2: SITE MARKERS & CLUSTERING ---
marker_cluster = MarkerCluster()
site_map.add_child(marker_cluster)

# Define color logic
def assign_marker_color(launch_outcome):
    return 'green' if launch_outcome == 1 else 'red'

spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# Add markers to cluster
for index, record in spacex_df.iterrows():
    marker = folium.Marker(
        location=[record['Lat'], record['Long']],
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=record['Launch Site']
    )
    marker_cluster.add_child(marker)

# --- 4. TASK 3: DISTANCE TO PROXIMITIES (Example: Coastline) ---
# Replace these with coordinates found using the MousePosition tool
launch_site_lat, launch_site_lon = 28.563197, -80.576820
coastline_lat, coastline_lon = 28.56367, -80.56763

distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# Add Distance Marker
distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html=f'<div style="font-size: 12; color:#d35400;"><b>{distance_coastline:.2f} KM</b></div>',
    )
)
site_map.add_child(distance_marker)

# Add Polyline
lines = folium.PolyLine(locations=[[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]], weight=1)
site_map.add_child(lines)

# --- DISPLAY FINAL MAP ---
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Define coordinates for the site and proximities
launch_site = [28.56319, -80.57682]
proximities = {
    'Highway': [28.56335, -80.57085],
    'Railway': [28.57206, -80.58521],
    'City': [28.61200, -80.80788]
}

for name, prox_coord in proximities.items():
    # 1. Calculate distance
    dist = calculate_distance(launch_site[0], launch_site[1], prox_coord[0], prox_coord[1])
    
    # 2. Add Marker for the proximity point
    distance_marker = folium.Marker(
        prox_coord,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12; color:#2574ad;"><b>{name}: {dist:.2f} KM</b></div>',
        )
    )
    site_map.add_child(distance_marker)
    
    # 3. Draw PolyLine between launch site and proximity
    line = folium.PolyLine(locations=[launch_site, prox_coord], weight=1, color='blue')
    site_map.add_child(line)

# Display final map
site_map
```
